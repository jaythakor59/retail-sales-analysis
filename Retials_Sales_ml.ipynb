```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv(r'Retails_Sales.csv')
```

```python
data.head(2)
```

```python
data.describe()
```

```python
data.info()
```

```python
x = data[['sales', 'quantity', 'discount', 'region', 'category', 'segment', 'ship_mode']]
```

```python
y = data['profit']
```

```python
x = pd.get_dummies(x,columns=['region','category','segment','ship_mode'],drop_first=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.metrics import r2_score , confusion_matrix , mean_squared_error
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train_scled = scaler.fit_transform(x_train)
```

```python
x_test_scled = scaler.transform(x_test)
```

```python
lr_model = LinearRegression()
```

```python
lr_model.fit(x_train_scled,y_train)
```

```python
ypred = lr_model.predict(x_test_scled)
```

```python
r2_score(ypred,y_test)
```

```python
np.sqrt(mean_squared_error(ypred,y_test))
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf_model = RandomForestRegressor()
```

```python
rf_model.fit(x_train_scled,y_train)
```

```python
rpred = rf_model.predict(x_test_scled)
```

```python
print("Random Forest R² Score:", r2_score(y_test, rpred))
```

```python
print("Random Forest RMSE:", np.sqrt(mean_squared_error(y_test, rpred)))
```
