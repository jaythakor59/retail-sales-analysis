```python
retails_Sales = retails_Sales.reset_index(drop=True)
```

```python
retails_Sales.isnull().sum()
```

```python
retails_Sales['ship_mode'] = retails_Sales['ship_mode'].fillna(retails_Sales['ship_mode'].mode()[0])
```

```python
retails_Sales['ship_mode'] = retails_Sales['ship_mode'].str.strip().str.title()
```

```python
retails_Sales['segment'] = retails_Sales['segment'].fillna(retails_Sales['segment'].mode()[0])
```

```python
retails_Sales['segment'] = retails_Sales['segment'].str.strip().str.title()
```

```python
retails_Sales['city'] = retails_Sales['city'].fillna('unkown')
```

```python
retails_Sales['region'] = retails_Sales['region'].fillna(retails_Sales['region'].mode()[0])
```

```python
retails_Sales['region'] = retails_Sales['region'].str.title()
```

```python
retails_Sales['discount'] = retails_Sales['discount'].fillna(retails_Sales['discount'].mean())
```

```python
retails_Sales['profit'] = retails_Sales['profit'].fillna(retails_Sales['profit'].mean())
```

```python
retails_Sales['order_date'] = pd.to_datetime(retails_Sales['order_date'],format='mixed')
```

```python
retails_Sales['ship_date'] = pd.to_datetime(retails_Sales['ship_date'])
```

```python
retails_Sales.to_csv('Retails_Sales.csv',index=False)
```
